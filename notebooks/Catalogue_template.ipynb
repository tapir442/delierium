# Catalogue template

An equation of the symmetry catalogue (`tests/symmetry_catalog.py`), its Janet basis and
pictures from `delierium.visualization`. Set `NAME` and run all cells.

```python
import sys
from pathlib import Path

from IPython.display import HTML, display
from sympy import Eq, symbols

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
sys.path.insert(0, str(ROOT))  # the catalogue lives in tests/
from tests.symmetry_catalog import CATALOG
from tests.test_symmetry_catalog import janet_basis, with_random_parameters

from delierium import ltf
from delierium.visualization import animate_flow, ode_solutions, staircase, vector_fields
```

```python
NAME = "Kamke 6.57, r = 3"  # all names: [e.name for e in CATALOG]
RANDOM_PARAMETERS = False  # random integers for the parameters: faster, generic dimension
VALUES = {}  # numbers for the parameters in the pictures, e.g. {"a": 2}; others are 1

entry = next(e for e in CATALOG if e.name == NAME)
indep, dep = entry.variables()
coordinates = symbols(entry.independent + entry.dependent)
for e in entry.parsed_equations():
    ltf(Eq(e, 0), dep, indep)  # in the notation of Lie: y'', u_xt
jb = janet_basis(with_random_parameters(entry) if RANDOM_PARAMETERS else entry)
print("dimension:", jb.type().dimension, "  catalogue:", entry.dimension)
```

```python
staircase(jb, names={"H": entry.dependent[0]}, title=entry.name);
```

```python
generators = entry.parsed_generators()
curves = ode_solutions(entry.parsed_equations()[0], dep[0], values=VALUES) if entry.kind == "ode" else []
axes = (1, 2) if entry.kind == "odes" else (0, 1)  # which two coordinates to draw
if generators:
    vector_fields(generators, coordinates, curves, axes=axes, values=VALUES);
```

```python
if generators and curves:
    display(HTML(animate_flow(generators[-1], coordinates, curves, values=VALUES).to_jshtml()))
```
