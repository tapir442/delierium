```python
# Example 1, p.136
x = Symbol('x')
u = Function('u')(x)
F = Function('F')
ode = diff(u, x) - F(u, x)
ltf(ode, [u], [x])
inf = prolongation(ode, create_infinitesimals(u, x), [u], [x])
ltf(inf, [u], [x])
inf = overdetermined_system_ode(ode, [u], [x])
print("determining system:")
for _ in inf:
    ltf(_, [u], [x])
```
