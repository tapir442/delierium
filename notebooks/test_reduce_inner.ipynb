```python
from sympy import *
from delierium.JanetBasis import LHDP, reduceS
from delierium.matrix_order import Context
```

```python
x, y = symbols("x y")
z = Function("z")(x, y)
w = Function("w")(x, y)
ctx = Context([w, z], [x, y])
f4 = LHDP(diff(w, x, y) + diff(z, x, y) + diff(w, y)/(2*y) - diff(w, x)/y + x*diff(z, y)/y - w/(2*y**2), ctx)
f1 = LHDP(diff(w, y) + x*diff(z, y)/(2*y*(x**2 + y)) - w/y, ctx)
f2 = LHDP(diff(z, x, y) + y*diff(w, y)/x + 2*y*diff(z, x)/x, ctx)
r = reduceS(f4, [f1, f2], ctx)
```
