```python
def difftotal(expr: Expr, diffby:Expr, diffmap: dict[Expr, Expr]) -> Expr:
    """Take the total derivative with respect to a variable.

    Example:

        theta, t, theta_dot = symbols("theta t theta_dot")
        difftotal(cos(theta), t, {theta: theta_dot})

    returns

        -theta_dot*sin(theta)
    """
    # stolen from https://robotfantastic.org/total-derivatives-in-sympy.html
    # Replace all symbols in the diffmap by a functional form
    fnexpr = expr.subs({s:s(diffby) for s in diffmap})
    # Do the differentiation
    diffexpr = diff(fnexpr, diffby)
    # Replace the Derivatives with the variables in diffmap
    derivmap = {Derivative(v(diffby), diffby):dv 
                for v,dv in diffmap.iteritems()}
    finaldiff = diffexpr.subs(derivmap)
    # Replace the functional forms with their original form
    return finaldiff.subs({s(diffby):s for s in diffmap})
```

```python
def tot_diff(e, x, y, order):
    res = []
    for i in range(1, order+1):
        du1 = symbols('du1')
        import pdb; pdb.set_trace()
        z2 = e.subs({diff(y, (x, i)) : du1})
        z2 = z2.diff(du1)
        z2 = z2.subs({du1 : diff(y, (x,i))})
        res.extend([diff(e, (x, i)), z2])
    return sum(res)
        
```

```python
def gamma(n, X, Y, x, y, dummy=Function('dummy')):
    U = X(x,y)*diff(dummy, x) + Y(x,y)*diff(dummy, y)
    if n == 0:
        return U
    if n == 1:
        return tot_diff(X(x,y), x, y, n-1) - diff(y, x) * tot_diff(Y(x,y), x, y, n-1)
    return tot_diff(gamma(n-1, X, Y, x, y, dummy), x, y, n-1) - diff(y, (x, n))*tot_diff(Y(x,y),x ,y, 0)
```

```python
def zeta(n, xi, eta, x, y):
    if n == 1:
        return tot_diff(eta, x, y, 1) + diff(y(x), x)*tot_diff(xi, x, y, 1)
    return tot_diff(zeta(n-1, xi, eta, x, y), x, y, n-1) - diff(y(x), (x,)*n)*tot_diff(xi, x, y, 1)
```
