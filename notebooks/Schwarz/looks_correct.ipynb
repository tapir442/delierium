```python
from sympy import * 
from IPython.core.debugger import set_trace
from pprint import pprint
```

```python
def order_of_expression(e, dependent, independent):
    for i in preorder_traversal(e):
        print(f"{i=}")
    return max([i.args[1][1] for i in preorder_traversal(e)
                if i.is_Derivative and 
                i.args[0] == dependent(independent)], default=1)
```

```python
def Dt(e, x, y, order=None):
    #set_trace()
    print("......")
    if order == None:
        order = order_of_expression(e, y, x)
    print(f"{e=}, {order=}")
    res = []
    for i in range(1, order+1):
        dummy=Dummy()
        print(f"{locals()=}")
        print(f"{Derivative(y(x), (x,i))=}")
        set_trace()
        z2 = e.subs({Derivative(y(x), (x, i)) : dummy})
        z2 = e.diff(diff(y(x), x, i))
        z2 = z2.subs({dummy : Derivative(y(x), (x, i))}).doit()
        res.extend([Derivative(e, (x, i)), z2])   
    res = sum(res)
    return res
```

```python
x = symbols('x', real=True)
y = symbols('y', cls=Function, real=True)
X=Function('X')
Y=Function('Y')
U=Function('U')
```

```python
vv=Dt(U, x=x, y=y).expand(); vv
```

```python
def zeta(X, Y, dep, indep, n):
    if n == 1:
        return Dt(Y(indep,dep(indep)), x=indep, y=dep, order=n) - Derivative(dep, indep, 1) * Dt(X(indep,dep(indep)), x=indep, y=dep, order=n)
    else:
        return Dt(zeta(X, Y, dep=dep, indep=indep, n=n-1), x=indep, y=dep, order=n-1) - Derivative(dep(indep), (indep, n)) * Dt(X, x=indep, y=dep, order=1)
```
