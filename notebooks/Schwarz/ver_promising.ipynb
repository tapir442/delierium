```python
%matplotlib notebook
from sympy import *
import os
os.environ["USE_SYMENGINE"] = "0"
from sympy.core.backend import *
init_printing()

f = symbols('f', cls=Function)
x = symbols('x', real=True)
y = symbols('y', real=True, cls=Function)
```

```python
f = sqrt(y(x)**3+(y(x).diff(x))**2)
f
```

```python
secondTerm = f.diff(Derivative(y(x) ,x,x))
secondTerm
```

```python
thirdTerm = f.diff(x)
thirdTerm
```

```python
fourthTerm = diff(thirdTerm, y(x))
fourthTerm
```

```python
fifthTerm=diff(fourthTerm*f, y(x), x)
fifthTerm
```

```python
%timeit diff(fourthTerm*f, y(x), x,x,x,x,x,y(x))
```
