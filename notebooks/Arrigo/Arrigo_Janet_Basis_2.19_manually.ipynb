```python
import os
#os.environ["USE_SYMENGINE"] = "0"

#from sympy.core.backend import *
from delierium.matrix_order import Context, Mgrlex, Mgrevlex, Mlex
from delierium.JanetBasis import Janet_Basis, LHDP
from delierium.helpers import ltf

from sympy import *

x = Symbol('x')
y = Symbol('y')
X = Function('X')(x, y)
Y = Function('Y')(x, y)

D = Derivative

s = [ D(X, (y, 2)),
    D(Y, (x, 2)) + (-y**4*x) * D(Y, y) + y * D(Y, x) + (2*y**4*x) * D(X, x) + (4*y**3*x) * Y + (y**4) * X,
    D(Y, y, x) + D(Y, x, y) + (-1) * D(X, (x, 2)) + (3*y**4*x) * D(X, y) + Y * D(X, x) + Y,
    D(Y, (y, 2)) + (-1) * D(X, y, x) + (-1) * D(X, x, y) + (2*y) * D(X, y)]
for _ in s:
    ltf(_, {X, Y}, [x, y])

```
