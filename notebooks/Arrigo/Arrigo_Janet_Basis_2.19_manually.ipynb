```python
import os
#os.environ["USE_SYMENGINE"] = "0"

from sympy import diff, symbols, Function, Symbol, Derivative
from delierium.matrix_order import Context, Mgrlex, Mgrevlex, Mlex
from delierium.JanetBasis import Janet_Basis, LHDP
from delierium.helpers import ltf, make_infinitesimal
from delierium.Infinitesimals import overdetermined_system_ode, janet_basis_from_ode



x = Symbol('x')
y = Symbol('y')
X = Function('X')(x, y)
Y = Function('Y')(x, y)

D = Derivative


# Example 2.19
x   = Symbol('x')
y   = Function('y')(x)
ode = y.diff(x,x) + 3*y*y.diff(x)+ y**3
independents = [x]
dependents = [y]

infinitesimals = {
    x: make_infinitesimal(x, x, y, name="X"),
    y: make_infinitesimal(y, x, y, name="Y"),
}

inf = overdetermined_system_ode(ode, [y], [x], infinitesimals)

s = [_ for _ in inf]
for _ in s:
    ltf(_, [X, Y, y], [x])

```
