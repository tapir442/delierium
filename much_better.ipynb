```python
def order_of_expression(e, dependent, independent):
    return max([i.args[1][1] for i in preorder_traversal(e)
                if i.is_Derivative and 
                i.args[0] == dependent], default=1)
```

```python
def prolongation(n):
    x=symbols('x')
    y=Function('y')(x)
    v=[x,y]+[diff(y, x, i) for i in range(1,n+1)]
    X=Function('X')(*v) # our dummy function

    print(f"{X=}")
    
    def Dt(order):        
        #set_trace()
        res = [Derivative(X , x)]
        for i in range(1, order+1):
            g1 = Derivative(y, x, i)
            g2 = Derivative(X, y, i)
            print(f"{g1=}")
            print(f"{g2=}")
            res.append(g1*g2)
        print(f"Dt{order}:{sum(res)=}")
        return sum(res)
    def my_zeta(n):
        print(f"my_zeta: {Dt(n)}")
        if n == 1:
            return Dt(n+1).subs({X:eta}) - diff(y, x)*Dt(n).subs({X:xi})
        return (Dt(n).subs({X: Dt(n-1).subs({X: my_zeta(n-1)})}) - 
             diff(y,x,n) * Dt(n).subs({X:xi}))
    return my_zeta(n)*Derivative(X, diff(y,x,n))
```

```python
prolongation(1)
```
