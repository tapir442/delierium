```python
def order_of_expression(e, dependent, independent):
    return max([i.args[1][1] for i in preorder_traversal(e)
                if i.is_Derivative and 
                i.args[0] == dependent], default=1)
```

```python
@cache
def Dt(x, y, order):
    set_trace()
    dummy = Function('dummy')(x, y)
    res = [dummy.diff(x)]
    print(f"{res=}")
    dummy=Dummy()
    for i in range(1, order+1):
        g1 = dummy.diff(x, order)
        g2 = dummy.diff(y, order-1)
        print(f"{g1=}")
        print(f"{g2=}")
        res.append(g1*g2)
    res = sum(res)
    return res
```

```python
def gurri(indep,dep,order):
    x,y=symbols('x, y')
    f=Function('f')(x)
    X=Function('X')(f, x)
    res = [Derivative(X , x)]
    for i in range(1, order+1):
        g1 = Derivative(f, x, i)
        g2 = Derivative(X, f, i)
        res.append(g1*g2)
    return X, sum(res).subs({f: dep, x: indep})
```

```python
gurri(x,y,2)
```

```python
def psi(n, e1, e2, x, y):
    if n == 1:
        rep, e = gurri(x,y,1)
        print(f"{e=}")
        print(f"{rep=}")
        print(f"{e1=}")
        print(f"{e2=}")
        print(f"{e.subs({rep: e1})=}")
        print(f"{e.xreplace({rep: e1})}")
        print(f"{e.replace(rep: e1})}")
        vv = e.replace(rep, e1) - Derivative(y,x)*e.replace(rep, e2)
        return vv
```
