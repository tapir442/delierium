```python
def order_of_expression(e, dependent, independent):
    return max([i.args[1][1] for i in preorder_traversal(e)
                if i.is_Derivative and 
                i.args[0] == dependent], default=1)
```

```python
def prolongation(n):
    x = symbols('x')
    y = Function('y')(x)
    X=Function('X')(x, y)

    def Dt(order):        
        #set_trace()
        res = [Derivative(X , x)]
        for i in range(1, order):
            g1 = Derivative(y, x, i)
            g2 = Derivative(X, y.diff(x, i), i)
            res.append(g1*g2)
        print(sum(res))
        return sum(res)
    def my_zeta(n):
        print(Dt(n))
        if n == 1:
            return Dt(n+1).subs({X:eta}) - diff(y, x)*Dt(n).subs({X:xi})
        return (Dt(n).subs({X: Dt(n-1).subs({X: my_zeta(n-1)})}) - 
             diff(y,x,n) * Dt(n).subs({X:xi}))
    return my_zeta(n)
```
